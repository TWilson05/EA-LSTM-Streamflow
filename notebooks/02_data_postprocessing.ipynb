Load each model and compute the ensemble mean predictions for each

```python
import pandas as pd
from src.config import OUTPUT_DATA_DIR

ENSEMBLE_MEMBERS = 10
# model_architectures = ["phase-split", "topographic", "area", "baseline"]
model_architectures = ["phase-split"]

for model in model_architectures:
    model_dir = OUTPUT_DATA_DIR / model
    # sum the predictions across ensemble members
    for i in range(ENSEMBLE_MEMBERS):
        member_file = model_dir / f"{model}_preds_member_{i}.csv"
        no_glac_member_file = model_dir / f"{model}_preds_noglacier_member_{i}.csv"
        member_df = pd.read_csv(member_file, index_col=0)
        no_glacier_df = pd.read_csv(no_glac_member_file, index_col=0)
        if i == 0:
            ensemble_df = member_df.copy()
            no_glacier_ensemble_df = no_glacier_df.copy()
        else:
            ensemble_df += member_df
            no_glacier_ensemble_df += no_glacier_df

    # average the predictions across ensemble members
    ensemble_df /= ENSEMBLE_MEMBERS
    no_glacier_ensemble_df /= ENSEMBLE_MEMBERS

    # save the ensemble predictions
    ensemble_file = model_dir / f"{model}_preds_ensemble.csv"
    ensemble_df.to_csv(ensemble_file, index=False)
    no_glacier_file = model_dir / f"{model}_preds_noglacier_ensemble.csv"
    no_glacier_ensemble_df.to_csv(no_glacier_file, index=False)
```
